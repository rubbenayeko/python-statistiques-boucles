Exercice 1 : le total du panier

Écrivez total_panier(prix) qui reçoit une liste de prix et renvoie le total à payer.

```python
def total_panier(prices):
    total = 0
    for price in prices:
        total = total + price #ou total += price
    return total

print(total_panier([500, 1200, 300]))   # 2000
print(total_panier([])) 
```

Exercice 2 : le plus grand nombre

Écrivez maximum(liste) qui renvoie le plus grand nombre d’une liste, sans utiliser max().
Indice : gardez dans une variable le plus grand nombre vu jusqu’ici, et mettez-la à jour pendant la boucle. Par quelle valeur la faire commencer ?

```python
def maximum(nums):
    max_num =  nums[0]
    for i in nums:
        if i > max_num:
            max_num = i
    return max_num

print(maximum([3, 9, 2, 7]))    # 9
print(maximum([-5, -2, -8]))    # -2
```

Exercice 3 : les admis

Écrivez nombre_admis(notes) qui compte combien de notes sont supérieures ou égales à 10.


```python
def nombre_admis(notes):
    for note in notes:
        if note < 10:
            notes.remove(note)
    return len(notes)   
        
print(nombre_admis([12, 8, 15, 9, 10]))   # 3
```

Exercice 4 : les nombres pairs

Écrivez somme_pairs(nombres) qui renvoie la somme des nombres pairs seulement.
Indice : un nombre est pair si n % 2 == 0.

```python

def somme_pairs(nombres):
    somme = 0
    for n in nombres:
        if n % 2 == 0:
            somme = n+somme
    return somme
        

print(somme_pairs([1, 2, 3, 4, 5, 6]))   # 12
```

Exercice 5 : un mot précis

Écrivez contient_mot(mots, cible) qui renvoie True si cible est dans la liste de mots, False sinon, sans utiliser in pour la liste.


```python
def contient_mot(mots,cibles):
    for mot in mots:
        if mot==cibles:
            return True
    return False

print(contient_mot(["riz", "mil", "maïs"], "mil"))   # True
print(contient_mot(["riz", "mil", "maïs"], "blé"))   # False
print(contient_mot([], "mil")) #False
```

Exercice 6 : tous valides ?

Écrivez tous_positifs(nombres) qui renvoie True si tous les nombres sont strictement positifs, False dès qu’un nombre ne l’est pas.


```python
def tous_positifs(nombres):
    for n in nombres:
        if n <= 0:
            return False
    return True   

print(tous_positifs([4, 8, 1]))     # True
print(tous_positifs([4, -2, 1]))    # False
print(tous_positifs([]))            # True
print(tous_positifs([4, 0, 1]))    # False
```

Exercice 7 : conversion de températures

Écrivez en_fahrenheit(celsius) qui reçoit une liste de températures en °C et renvoie une nouvelle liste en °F (formule : c * 9/5 + 32).
Indice : commencez par une liste vide resultat = [] et utilisez resultat.append(...).

```python
def en_fahrenheit(celsius):
    resultat = []
    for c in celsius:
        resultat.append(c*9/5+32)
    return resultat 
print(en_fahrenheit([0, 100, 37]))   # [32.0, 212.0, 98.6...]
```

Exercice 8 : filtrer

Écrivez mots_longs(mots, n) qui renvoie la liste des mots ayant plus de n lettres.



```python
def mots_longs(mots, n):
    resultat = []
    for mot in mots:
        if len(mot) > n:
            resultat.append(mot)
    return resultat        
    
print(mots_longs(["chat", "éléphant", "rat", "girafe"], 4))   # ['éléphant', 'girafe']
```

Exercice 9 : moyenne des notes valides

Une liste contient des notes sur 20, mais certaines valeurs sont invalides (négatives ou supérieures à 20). Écrivez moyenne_valide(notes) qui calcule la moyenne en ignorant les notes invalides.


```python
def moyenne_valide(notes):
    valides =  []
    for note in notes:
        if 0 <= note <= 20:
           valides.append(note)
    if len(valides) == 0:
        return 0
    return sum(valides)/len(valides)

print (moyenne_valide([12, 15, -1, 18, 25, 9]))   # 13.5
```

Exercice 10 : compter les occurrences

Écrivez compter_lettre(texte, lettre) qui compte combien de fois une lettre apparaît dans un texte. Une chaîne se parcourt avec for comme une liste.


```python
def compter_lettre(texte, lettre):
    compteur = 0
    for letter in texte:
        if letter == lettre:
            compteur =compteur + 1
    return compteur        

print(compter_lettre("bonjour tout le monde", "o"))   # 4
```

Exercice 11 : boucle while

Un capital de 1000 est placé avec un intérêt de 5 % par an. Écrivez annees_pour_doubler(capital) qui renvoie le nombre d’années nécessaires pour que le capital dépasse le double de sa valeur de départ.
Indice : utilisez while et une variable annees qui augmente de 1 à chaque tour.


```python
def annees_pour_doubler(capital):
    taux = 0.05
    annee = 0
    objectif = capital * 2
    while capital < objectif:
        capital = capital * (1 + taux)
        annee = annee + 1 #ou annee += 1
    return annee

print(annees_pour_doubler(1000)) 



```

NB:La boucle au-dessus et celle en dessous semblent similaire mais ce n'est pas le cas: La première est réaliste et la seconde absurde. L'idéal est de toujours fixé l'objectif clairement avant la boucle afin d'éviter que cette dernière ne s'arrète plus. Cela va toujours engendrer une boucle infinie puisque dans ma boucle while, je suis entrain de comparer capital à lui-même (capital < capital * 2), ce qui sera toujours vérifiée d'où la boucle infinie.

TOUJOURS FIXER L'OBJECTIF CLAIREMENT AVANT LA BOUCLE.

```python
def annees_pour_doubler(capital):
    taux = 0.05
    annee = 0
    while capital < capital * 2:
        capital = capital * (1 + taux)
        annee = annee + 1 #ou annee += 1
    return annee

print(annees_pour_doubler(1000)) 
```
